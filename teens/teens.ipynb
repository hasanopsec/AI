### My Question
**Is there an association between a teenager's gender and their preferred life status (being happy, healthy, rich, or famous)?**

- **Variable 1 (Categorical):** `Gender` (Male, Female)
- **Variable 2 (Categorical):** `Preferred_Status` (Happy, Healthy, Rich, Famous)


### My Study
This is an **observational study** (analyzing survey sample data).

- **Why it is an observational study:**
  We are observing and recording pre-existing characteristics, personal preferences, and habits of students through self-reported survey responses. Researchers did not manipulate conditions, intervene in the students' lives, or impose any treatments. Because the data is observational, it can reveal **associations** between variables, but it **cannot prove cause-and-effect**.
- **Why it is not an experiment:**
  In a true experiment, researchers deliberately impose a **treatment** or manipulate an independent variable across randomized treatment and control groups to evaluate cause-and-effect. In this study, nobody assigned students to a gender or forced them into a preferred status; all variables were naturally occurring and non-manipulated.
- **Why it is not a census study (despite the name "Census at School"):**
  A census collects data from **every single individual in the entire target population**. While the project is titled *Census at School*, our dataset contains responses from a sample of only **500 students** out of the tens of millions of teenagers in the United States. Furthermore, participation was voluntary (teachers chose whether to participate). Because it only examines a subset of the population, it is a sample survey, not a census study.


### 1. Data Preparation & Two-Way Frequency Table


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.graphics.mosaicplot import mosaic

# Load dataset
df = pd.read_csv('CensusAtSchool.csv')

# Clean categorical variables
df_clean = df[['Gender', 'Preferred_Status']].dropna().copy()
df_clean['Gender'] = df_clean['Gender'].str.strip()
df_clean['Preferred_Status'] = df_clean['Preferred_Status'].str.strip()

# Two-way table of raw counts
counts = pd.crosstab(df_clean['Gender'], df_clean['Preferred_Status'])
print('--- Two-Way Table (Raw Counts) ---')
print(counts)

# Two-way table of conditional percentages (% within each gender)
props = pd.crosstab(df_clean['Gender'], df_clean['Preferred_Status'], normalize='index') * 100
print('\n--- Conditional Percentages (% within Gender) ---')
print(props.round(1))
```

### 2. Option A: Side-by-Side Bar Chart (`sns.countplot` / `sns.catplot`)
A side-by-side bar chart compares the count of each preferred status directly between males and females.

```python
# Side-by-side bar chart using Seaborn
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df_clean,
    x='Preferred_Status',
    hue='Gender',
    palette='Set2'
)
plt.title('Side-by-Side Bar Chart: Preferred Status by Gender')
plt.xlabel('Preferred Life Status')
plt.ylabel('Count of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 3. Option B: Segmented (Stacked) Bar Chart (`.plot(kind='bar', stacked=True)`)
A 100% segmented bar chart displays the relative proportions of each status within each gender group.

```python
# 100% Segmented (stacked) bar chart
ax = props.plot(kind='bar', stacked=True, figsize=(8, 5), colormap='Set2')
plt.title('Segmented Bar Chart: Preferred Status Breakdown by Gender (%)')
plt.xlabel('Gender')
plt.ylabel('Percentage (%)')
plt.legend(title='Preferred Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 4. Option C: Mosaic Plot (`statsmodels.graphics.mosaicplot.mosaic`)
A mosaic plot visualizes both the marginal distribution of gender (tile width) and conditional distributions of preferred status (tile height).

```python
# Mosaic plot using Statsmodels
fig, ax = plt.subplots(figsize=(9, 6))
mosaic(
    df_clean,
    ['Gender', 'Preferred_Status'],
    ax=ax,
    title='Mosaic Plot: Association between Gender and Preferred Status'
)
plt.tight_layout()
plt.show()
```

### Key Findings
- **Happiness is the top priority for both groups**, but significantly more **females (63.2%)** prioritize being *Happy* compared to **males (44.5%)**.
- **Wealth is prioritized more by males:** **34.1%** of males chose *Rich* as their preferred status compared to **20.1%** of females.
- **Fame is rare overall**, but chosen more often by males (6.6%) than females (2.6%).
- **Health was chosen at nearly identical rates** across both genders (~14%).

### Option 3: Mosaic Plot in Desmos (Hard!)
In a mosaic plot, **both dimensions represent probabilities**:
- **X-axis (Column Width):** Marginal proportion of Gender ($N = 445$)
  - Females = $234 / 445 = 52.6\%$ (width $0$ to $52.6$)
  - Males = $211 / 445 = 47.4\%$ (width $52.6$ to $100$)
- **Y-axis (Segment Height):** Conditional proportion of Preferred Status within that gender ($0\%$ to $100\%$)
- **Tile Area:** The area of each tile is proportional to the joint frequency of students!

**Desmos Equations for Mosaic Plot:**
- **Female Tiles ($x \in [0, 52.6]$):**
  - Happy ($0\% \to 63.2\%$): `polygon((0, 0), (52.6, 0), (52.6, 63.2), (0, 63.2))`
  - Rich ($63.2\% \to 83.3\%$): `polygon((0, 63.2), (52.6, 63.2), (52.6, 83.3), (0, 83.3))`

- **Male Tiles ($x \in [52.6, 100]$):**
  - Happy ($0\% \to 44.5\%$): `polygon((52.6, 0), (100, 0), (100, 44.5), (52.6, 44.5))`
