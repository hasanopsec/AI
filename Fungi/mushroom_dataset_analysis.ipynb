# Mushroom Dataset Analysis: Most Common Cap Color

## Study Description
This study investigates the distribution of mushroom cap colors using the UCI Mushroom dataset (`mushrooms.csv`). The primary objective is to identify the most frequently occurring cap color across all recorded mushroom specimens and examine how secondary attributes—specifically odor—are distributed within each color category.

**Dataset Overview:**
- **Feature Examined:** `cap-color` (e.g., `n` = brown, `g` = gray, `e` = red, `y` = yellow, `w` = white, `b` = buff, `p` = pink, `c` = cinnamon, `u` = purple, `r` = green)
- **Secondary Feature:** `odor` (e.g., `n` = none, `f` = foul, `y` = fishy, `s` = spicy, `a` = almond, `l` = anise, `p` = pungent, `c` = creosote, `m` = musty)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("mushrooms.csv")

# Directly compute and print the cap color totals
cap_color_totals = df["cap-color"].value_counts()
print("--- Cap Color Total Counts ---")
print(cap_color_totals)

# Prepare data for visualization
counts = (
    df[["cap-color", "odor"]]
    .value_counts()
    .reset_index(name="count")
)

# Generate the plot
plt.figure(figsize=(10, 5))
sns.barplot(
    data=counts,
    x="cap-color",
    y="count",
    hue="odor"
)

plt.title("Mushroom Cap Color Breakdown by Odor")
plt.xlabel("Cap Color Category")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.legend(title="Odor", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## Conclusion

- **Most Common Color:** The single most prevalent cap color in the dataset is **brown (`n`)** with **2,284 specimens**, followed closely by **gray (`g`)** with **1,840** and **red (`e`)** with **1,500**.
- **Distribution Insight:** Rare colors such as purple (`u`) and green (`r`) appear in negligible quantities (only 16 specimens each).
- **Odor Correlation:** Within the dominant brown (`n`) category, the largest share of mushrooms possesses no odor (`n`), though foul (`f`), spicy (`s`), and fishy (`y`) odors are also prominent across the dataset.

In summary, brown (`n`) is definitively the most common mushroom cap color out of all specimens evaluated in this dataset.
